```python
# średnia wieku
def avg_age(data_):
    total = 0
    for person in data_:
        total += person['age']
    return total/len(data_)

avg_age(data)
```

```python
# średnia wieku generator expression
def avg_age(data_):
    return sum(person['age'] for person in data_) / len(data_)

print(avg_age(data))
```

```python
%%timeit

# średnia wieku dla osób spoza warszawy
def avg_age_out_wwa(data_):
    total = 0
    counter =0

    for person in data_:
        if person['city'].lower() != 'warsaw':
            total += person['age']
            counter += 1
    return total // counter

avg_age_out_wwa(data)
```

```python
%%timeit

# średnia wieku dla osób spoza warszawy
def avg_age_out_wwa(data_):
    people_not_warsaw = [person['age'] for person in data_ if person['city'].lower() != 'warsaw']
    return sum(people_not_warsaw) // len(people_not_warsaw)

avg_age_out_wwa(data)
```

```python
%%timeit

# średnia wieku dla osób spoza Warszawy
def avg_age_out_wwa(data_):
    people_not_warsaw = tuple(map(lambda person: person['age'], filter(lambda person: person['city'].lower() != 'warsaw', data_)))
    return sum(people_not_warsaw) // len(people_not_warsaw)

avg_age_out_wwa(data)
```

```python
def map_(callback, collection):
    result = []

    for item in collection:
        result.append(callback(item))

    return result
```

```python
digits = [1, 2, 3, 4, 5]

map_(lambda digit: digit * 2, digits)
```

```python
def filter_(callback, collection):
    result = []

    for item in collection:
        if callback(item):
            result.append(item)

    return result

filter_(lambda digit: digit > 2, digits)
```

```python
# create hash (2 pierwsze znaki z kaΩdego values) ->
# hash piwa37jspydr


def create_hash(people):
    hashes = []
    for person in people:
        text = ''

        for val in person.values():
            if isinstance(val, list):
                for item in val:
                    text += item[:2]

            else:
                text += str(val)[:2]

        hashes.append(text.lower())
    return hashes


create_hash(data)
```

```python
# wersja od ai

def create_hash(people):
    return [
        ''.join(
            item[:2] if isinstance(val, list) else str(val)[:2]
            for val in person.values()
            for item in (val if isinstance(val, list) else [val])
        ).lower()
        for person in people
    ]
```
