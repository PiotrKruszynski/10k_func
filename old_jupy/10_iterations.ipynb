```python
# Średnia wieku na trzy sposoby
# to rozwizanie jest najszybsze ale ma jedn wadę. Jest imperatywne - mówimy krok po kroku co trzeba zrobić aby osiagnac ce

temp = 0
for person in data:
    temp += person['age']
temp / len(data)
```

```python
temp = 0
counter = 0

while counter < len(data):
    temp += data[counter]["age"]
    counter += 1
    
temp / len(data)
```

```python
# comprehension - pythonic way do tworzenia klekcji, w tym przypadku listy
# ta opcje wybieramy jezeli chcemy generowac kolekcje - z defaultu

avg = sum([person['age'] for person in data])/len(data)
print(avg)
```

```python
# to samo jako generator expression
avg = sum(person['age'] for person in data)/len(data)
print(avg)
```

```python
# Ta opcje wybieramy, jezeli piszemy w paradygmacie funkcyjnym
# paradygmat funkcyjny -> pochodzi od rachunku lambda

# map( jako pierwszy parametr przyjmuje callback (deklaracje funkcji -> najczęściej lambda by nie zasmiesmiecac przestrzeni kolejna zmienna), jako drugi iterable) -> lazy evaluation
# jest deklaratywny -> mówi co chcemy
map(lambda person: person["age"], data)
```

```python
sum(map(lambda person: person["age"], data)) / len(data)
```

```python
# powyzsze mozna tez tak:

def convert_dict(person):
    return person['age']

sum(map(convert_dict, data)) / len(data)
```

```python
# KOlejne cwiczenie. Sprawdz czy wszyscy maja w imieniu literke a , jezeli tak zwroc True, inaczej False
'e' in 'ela ma kota'
```

```python
def check_names_with_a_letter(persons):
    for person in persons:
        if 'a' not in person['name']:
            return False
    return True

check_names_with_a_letter(data)
```

```python
def checknames_with_a_letter2(items):
    return len([item for item in items if 'a' in item['name']]) == len(items)

checknames_with_a_letter2(data)
```

```python
def checknames_with_a_letter3(items):
    return len(list(filter(lambda item: 'a' in item['name'], items))) == len(items)

checknames_with_a_letter2(data)
```
