# Comprehensions

1. List comprehension

```python
[element for element in iterable]
[mapping/transformation for element in iterable]
[mapping for element in iterable if filter]

```

2. Set comprehension

```python
[element for element in iterable]
[mapping/transformation for element in iterable]
[mapping for element in iterable if filter]

```

4. Dict comprehension

```python
[key:val for keyt in iterable]
[mapping/transformation for element in iterable]
[mapping for element in iterable if filter]

```

6. General expression

```python
(element for element in iterable)
(mapping/transformation for element in iterable)
(mapping for element in iterable if filter)

```

## Generator expresssion

działa 1:1 tak jak list tylko że jest :
lazy evaluated - kalkulacje sa robione dopiero w momencie, kiedy potrzebujemy pobrać wartość. 
Domyślnie to nic nie robi, jest to coś w rodzaju wzoru jak wyliczyć wartosci.

map i filter sa też lazy evaluated

jak jest wrzucany do list() to nie potrzebuje podwójnych nawiasów!!

```python
(x for x in iterable)
(mapping for x in iterable if filter)

```

````python
rows, letters = seating_plan
[row for row in rows]
````

````python
# '_' - dummy name

seats = [{letter: None for letter in letters} for _ in rows]
seats
````
