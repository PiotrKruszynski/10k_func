# <font color="green">*RANGE*</font>
1. iterable o zadanych wartościach
2. uporządkowana
3. homogeniczny (INT)
4. mutowalna

```python
# Podstawowe użycie
r1 = range(5)        # 0, 1, 2, 3, 4
r2 = range(2, 5)     # 2, 3, 4
r3 = range(0, 10, 2) # 0, 2, 4, 6, 8 (krok 2)
r4 = range(5, 0, -1) # 5, 4, 3, 2, 1 (malejąco)
```

```python
r = range(10)
# Niemutowalny
# Zajmuje stałą ilość pamięci
# Generuje wartości na żądanie
print(len(r))        # 10
print(r.start)       # 0
print(r.stop)        # 10
print(r.step)        # 1
```

```python
r = range(5)
# Sprawdzanie zawartości
print(3 in r)        # True
print(10 in r)       # False

# Indeksowanie i wycinanie
print(r[0])          # 0
print(r[-1])         # 4
print(r[::2])        # range(0, 5, 2)

# Konwersja na listę
lista = list(r)      # [0, 1, 2, 3, 4]
```

```python
# praktyczne zastosowanie
# Dict z range
dict_from_range = {x: x**2 for x in range(5)}
print(dict_from_range)  # {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

# Range w pętlach
for i in range(5):
    print(i, end=' ')  # 0 1 2 3 4

# Iterowanie przez indeksy
lista = ['a', 'b', 'c']
for i in range(len(lista)):
    print(f"Index {i}: {lista[i]}")

# Tworzenie sekwencji liczbowych
parzyste = list(range(0, 11, 2))  # [0, 2, 4, 6, 8, 10]
```

```python
# przykłady zaawansowane
# Słownik z dynamicznym generowaniem wartości
d = {x: [y for y in range(x)] for x in range(5)}
print(d)  # {0: [], 1: [0], 2: [0, 1], 3: [0, 1, 2], 4: [0, 1, 2, 3]}

# Zagnieżdżone pętle z range
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={j}")

# Lista krotek z range
coords = [(x, y) for x in range(2) for y in range(2)]
print(coords)  # [(0, 0), (0, 1), (1, 0), (1, 1)]
```
