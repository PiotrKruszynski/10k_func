# Guess Game

1. Użytkownik podaje minim zakresu
2. Użytkownik podaje maximm zakresu
3. Komputer wylicza losową liczbę z zakresu
4. Użytkownik zgaduje losową liczbę:
    - jak za mało to mówi
    - jak za dużo to mówi
    - jak zgadł to mówi za którym razem
  
$y = sin(\frac{x^2}{2})$

```python
min_digit = int(input('Podaj minimum zakresu:\n'))
max_digit = int(input('Podaj maximum zakresu:\n'))
```

```python
def get_user_digits():
    min_digit = int(input('Podaj minimum zakresu:\n'))
    max_digit = int(input('Podaj maximum zakresu:\n'))
    return min_digit, max_digit
```

PKT 3 -> KOMPUTER LOSUJE LICZBĘ:

```python
from random import randint
def get_random_digit(min_d, max_d):
    return randint(min_d, max_d)
get_random_digit(3, 10)
```

Pkt 4 -> Użytkownik zgaduje losową liczbę

```python

def guess_digit(drawn_digit, min_digit, max_digit):
    print('Zgadnij liczbę losową z zakresu {min_digit}-{max_digit}:')
    counter = 0
    
    while True:
        counter += 1
        user_digit = int(input())
    
        if user_digit < drawn_digit:
            print('Za mało, spróbuj jeszcze raz')
        elif user_digit > drawn_digit:
            print('Za dużo, spróbuj jeszcze raz')
        else:
            print(f'Trafileś. Zgadłeś za {counter} razem')
            break
            
```

specjalna składnia z przecikiem, żeby odebrać faukcje get_user_digit

```python
def play():
    min_digit, max_digit = get_user_digits()
    shuffled_digit = get_random_digit(min_digit, max_digit)
    guess_digit(shuffled_digit, min_digit, max_digit)

    if input('Czy chcesz grać ponownie?[y/n]\n').lower() == 'y':
        play()

    print('Game Over')
```

```python
play()
```
