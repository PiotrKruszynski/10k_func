```python
# Na podstawie zdobytych umiejętności napisz funkcję UUID (universal unique id)
# każde wywołanie funkcji ma zwracać kolejną liczbę całkowitą

# gen() -> 1
# gen() -> 2
# gen() -> 3
```

```python
def gen_uuid():
    x = 1
    def inner():
        nonlocal x
        result = x
        x = x + 1
        return result
    return inner
```

```python
# dodatkowa wersja z licznikiem
def gen_uuid(idx=1):
    def inner():
        nonlocal idx
        result = idx
        idx = idx + 1
        return result
    return inner
```

```python
# dodatkowa wersja z licznikiem z możliwością resetu
def gen_uuid(idx = 1):
    def inner(new_idx = None):
        nonlocal idx

        if new_idx is not None:
            idx = new_idx

        result = idx
        idx += 1
        return result
    return inner
```

# Decorators

1. Dekoratory funkcyjne (robione funkcją/metodą)
2. Dekoratory class -> zalecany sposób aby nie stosować metaklas
3. Design pattern

DDD - Dekorator Driven Development

Ad.1

Dekorator to funkcja, która dodaje lub modyfikuje działanie innej funkcji.


```python
#schemat

def upper(fn):
    def inner(name):
        return fn(name).title()

    return inner
```

```python
#dekoratory nakłada się na deklarację funkcji
```

```python
@upper
def hello(name):
    return name
```

```python
hello('andrzej')
```

### Higher order function
- (Funkcja wyższego rzędu) - funkcja, która przyjmuje jako parametr, bądź zwraca deklarację funkcji
- Aby tak było funkcje w języku muszą być 'first class functions, first class citizens' ( inaczej muszą być obiektami)
- Możemy osiągnąć kompozycje -> nakładanie funkcji na siebie

```python
def upper(text):
    return text.title()

def reverse(text):
    return text[::-1]

def sentence(fn, text):
    return fn(text)
```

```python
sentence(reverse, 'ala ma kot i wszy')
```

```python
sentence(upper, 'ala ma kot i wszy')
```

```python
sentence(reverse, sentence(upper, 'ala ma kot i wszy'))
```
