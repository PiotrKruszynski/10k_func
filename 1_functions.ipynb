## Parametry / Argumenty
- delaracja - parametry
- wywołanie - argumenty

```python
def magic(a, b, c, d, e):
    pass
```

```python
# argumenty pozycyjne
magic(1, 2, 3, 4, 5)

# argumenty nazwane
magic(a=1, b=2, c=3, d=4, e=5)
magic(b=2, d=4, e=5, a=1, c=3)

# kombinacja pozycyjne + nazwane (pozycyjne zawsze przed nazwanymi)
magic(1, 2, 3, d=4, e=5)
magic(1, 2, 3, e=4, d=5)
```

```python
def magic(a, b=None):
    pass
```

```python
# optional/ default (zawsze po pozycyjnych)
magic(1)
magic(1, 2)
```

```python
# default używać tylko niemodyfikowalnych (immutable) typów danych -> najczęściej w python używa None bądź primitive type
def magic_pro(item, array=[]):
    array.append(item)
    return array

# koszerna wesja
def magic_pro(item, array=None):
    if array is None:
        array = []
    array.append(item)
    return array
```

```python
magic_pro(42)
```

```python
magic_pro(1, [])
```

```python
magic_pro(666)
```

```python
# zmienna ilości parametrów
def magic(*args, **kwargs):
    print(args)
    print(type(args))
    print(kwargs)
    print(type(kwargs))
```

```python
magic()
```

```python
magic(1, 2, 3)
```

```python
magic(a=1, b=2, c=3)
```

```python
magic(1, 2, 3, b=4, c=5)
```

```python
# / przed nim tylko pozycyjne
# * po niej tylko nazwane

def magic(a, b, /, c, *, d, e):
    pass
```

```python
magic(1,2,3,d=1,e=1)
magic(1,2,c=3,d=1,e=1)
```
